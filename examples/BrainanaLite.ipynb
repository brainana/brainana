```python
# Check T1w NIfTI(s) — fail fast before env install. Colab+Drive: mounts Drive and waits for sync.
import gzip
import os
import shutil
import time
from pathlib import Path

try:
    import google.colab  # noqa: F401
    IS_COLAB = True
except ImportError:
    IS_COLAB = False

# Mount Drive BEFORE any mkdir under /content/drive: creating WORKING_DIR while Drive is
# unmounted would leave a fake local tree that later masquerades as a mount.
ON_DRIVE_FUSE = IS_COLAB and str(WORKING_DIR).startswith("/content/drive")
if ON_DRIVE_FUSE:
    from google.colab import drive
    if os.path.ismount("/content/drive"):
        print("Google Drive already mounted.")
    else:
        # /content/drive exists but is NOT a real mount => stale local folder created by
        # writing there before mounting. ismount() is False => safe to clear (not real Drive).
        if Path("/content/drive").exists():
            print("Clearing stale local /content/drive (was not a real Drive mount)…")
            shutil.rmtree("/content/drive")
        print("Connecting Google Drive via Colab — complete Google's sign-in / allow prompt, then return here.")
        drive.mount("/content/drive")
    if not os.path.ismount("/content/drive"):
        raise RuntimeError("Google Drive mount did not complete — re-run this cell and finish the sign-in/allow prompt.")

WORKING_DIR = Path(WORKING_DIR)
WORKING_DIR.mkdir(parents=True, exist_ok=True)  # now safe: real Drive, not a local stub

DRIVE_NIFTI_TIMEOUT = 120  # seconds to wait for Drive upload + FUSE hydration
DRIVE_NIFTI_POLL = 3
DEMO_NIFTI = "exam_T1w_ple.nii.gz"  # fetched during env setup when RUN_DEMO=True

def find_t1w_niftis(root: Path) -> list[Path]:
    """Top-level .nii/.nii.gz under root (non-recursive)."""
    if not root.is_dir():
        return []
    return sorted(p for p in root.iterdir() if p.is_file() and (p.name.endswith(".nii.gz") or p.name.endswith(".nii")))

def wait_for_real_nifti(path, *, timeout=DRIVE_NIFTI_TIMEOUT, poll=DRIVE_NIFTI_POLL):
    """Force FUSE hydration and verify the file is a readable NIfTI (not a placeholder)."""
    path = Path(path)
    deadline = time.monotonic() + timeout
    last = None
    while time.monotonic() < deadline:
        try:
            with open(path, "rb") as f:
                head = f.read(2)
            if len(head) < 2:
                last = "zero/short read (not hydrated yet)"
            elif path.name.endswith(".nii.gz"):
                if head != b"\x1f\x8b":
                    last = f"bad gzip magic {head!r} (placeholder/partial sync?)"
                else:
                    with gzip.open(path, "rb") as g:
                        g.read(352)  # NIfTI-1 header; raises if truncated
                    return
            else:
                with open(path, "rb") as f:
                    hdr = f.read(348)
                if len(hdr) < 348:
                    last = "truncated uncompressed NIfTI"
                else:
                    return
        except OSError as exc:
            last = f"read error: {exc}"
        except Exception as exc:
            last = str(exc)
        print(f"  …waiting for file to sync: {path.name} ({last})")
        time.sleep(poll)
    raise TimeoutError(f"{path} never became a valid NIfTI ({last}). On Colab+Drive: wait for Drive Desktop 'Sync complete' before Run All.")

def discover_t1w_candidates() -> list[Path]:
    candidates = find_t1w_niftis(WORKING_DIR)
    legacy = WORKING_DIR / "input_T1w"
    if not candidates and legacy.is_dir():
        candidates = find_t1w_niftis(legacy)
    return candidates

candidates = discover_t1w_candidates()

if RUN_DEMO:
    # The example volume this notebook downloads is not a conflict. Exempting it is what
    # makes RUN_DEMO re-runnable: it lands in WORKING_DIR, so on a second Run All the old
    # check aborted on the very file the first run created.
    stray = [p for p in candidates if p.name != DEMO_NIFTI]
    if stray:
        names = ", ".join(p.name for p in stray)
        raise RuntimeError(f"RUN_DEMO=True but found NIfTI(s) in {WORKING_DIR}: {names}. Demo mode requires no .nii/.nii.gz files in WORKING_DIR (top level or input_T1w/) other than {DEMO_NIFTI}. Remove them or set RUN_DEMO=False.")
    if candidates:
        print(f"RUN_DEMO: reusing the example T1w already in {WORKING_DIR} ({DEMO_NIFTI}).")
    else:
        print("RUN_DEMO: example T1w will be fetched after environment setup.")
elif not candidates:
    listing = sorted(p.name for p in WORKING_DIR.iterdir()) if WORKING_DIR.is_dir() else []
    msg = [f"No T1w .nii/.nii.gz files found in {WORKING_DIR}.",
           f"Folder currently contains: {listing or '(empty)'}."]
    if ON_DRIVE_FUSE:
        msg.append("Drive is mounted but this folder has no NIfTIs — check that "
                   "(1) the mounted Google account holds the files, "
                   "(2) WORKING_DIR points at the exact subfolder, "
                   "(3) Drive shows 'Sync complete'.")
    msg.append("Or set RUN_DEMO=True to fetch an example volume.")
    raise FileNotFoundError(" ".join(msg))
else:
    print(f"Checking T1w input ({len(candidates)} file(s))...")
    for p in candidates:
        wait_for_real_nifti(p)
        print(f"  ✓ {p.name}")
    print(f"T1w input OK — ready in {WORKING_DIR}")
```

```python
# Environment setup — clone, install deps. No edits needed.
# First Colab run may restart the runtime when install finishes; Run All again.

import os
import re
import shutil
import subprocess
import threading
import sys
import logging
import importlib.util
import time
import urllib.request
from pathlib import Path

if "find_t1w_niftis" not in globals():
    raise RuntimeError("Run the 'Check T1w NIfTI(s)' cell above first.")

_step_times: dict[str, float] = {}
_step_stack: list[str] = []


def _format_duration(seconds: float) -> str:
    if seconds < 60:
        return f"{seconds:.1f}s"
    minutes, secs = divmod(seconds, 60)
    if minutes < 60:
        return f"{int(minutes)}m {secs:.0f}s"
    hours, minutes = divmod(minutes, 60)
    return f"{int(hours)}h {int(minutes)}m {secs:.0f}s"


def _step_indent(depth: int | None = None) -> str:
    if depth is None:
        depth = max(0, len(_step_stack) - 1)
    return "  " * depth


def _step_detail_lines(detail: str, indent: str) -> list[str]:
    if not detail:
        return []
    sub = indent + "  "
    return [sub + line if line else "" for line in detail.splitlines()]


def step_log(name: str, *, done: bool = False, detail: str = "") -> None:
    """Print step start/finish with hierarchy, icons, and elapsed time."""
    if done:
        duration = ""
        if name in _step_times:
            duration = _format_duration(time.perf_counter() - _step_times.pop(name))
        depth = max(0, len(_step_stack) - 1)
        if _step_stack and _step_stack[-1] == name:
            _step_stack.pop()
        indent = _step_indent(depth)
        line = f"{indent}✓ {name}"
        if duration:
            line += f"  ·  {duration}"
        lines = [line, *_step_detail_lines(detail, indent)]
        print("\n".join(lines), flush=True)
        if depth == 0 and re.match(r"^Step [1-9]", name):
            print(flush=True)
    else:
        if not _step_stack and re.match(r"^Step [1-9]", name):
            print(flush=True)
        depth = len(_step_stack)
        _step_stack.append(name)
        _step_times[name] = time.perf_counter()
        print(f"{_step_indent(depth)}▶ {name}", flush=True)


def step_slow_hint() -> None:
    """Note before preprocessing steps that can run for minutes."""
    print(f"{_step_indent()}  This step may take a while.", flush=True)


BRAINANA_URL  = "https://github.com/brainana/brainana.git"
FIREANTS_URL  = "https://github.com/rohitrango/FireANTs.git"
FIREANTS_REF  = "v1.5.0"   # fused_ops built from source; base fireants comes from brainana deps

# Bound every network call: a notebook has no stdin, so a git prompt or a stalled socket
# hangs the cell forever with no output rather than failing.
CLONE_TIMEOUT = 1800   # git clone (the repo is large)
INSTALL_TIMEOUT = 3600 # uv/pip install (builds from source on some platforms)
DOWNLOAD_TIMEOUT = 120 # per-read socket timeout for the demo volume
INSTALL_HEARTBEAT = 30 # seconds between "still working" notes while uv is quiet

try:
    import google.colab  # noqa: F401
    IS_COLAB = True
except ImportError:
    IS_COLAB = False

# Local: warn on system/base Python (installs affect this kernel).
_in_venv = sys.prefix != sys.base_prefix
_conda_env = os.environ.get("CONDA_DEFAULT_ENV", "")
ISOLATED_ENV = _in_venv or (_conda_env not in ("", "base"))
if (not IS_COLAB) and (not ISOLATED_ENV):
    print("WARNING: this kernel runs in a non-isolated (system/base) Python; installs here may")
    print("         pollute it. Recommended — launch Jupyter from an isolated env first, e.g.:")
    print("           uv venv .venv && source .venv/bin/activate \\")
    print("             && uv pip install jupyterlab && jupyter lab")
    print("         Continuing anyway.\n")

WORKING_DIR = Path(WORKING_DIR)
BIDS_DIR   = WORKING_DIR / "input_bids"           # generated BIDS
OUTPUT_DIR = WORKING_DIR / "brainana_lite_output" # outputs / work / QC


def copy_nifti_validated(src: Path, dst: Path) -> None:
    """Drive FUSE: wait for hydration, then gzip-safe copy (see create_output_link)."""
    wait_for_real_nifti(src)
    dst.parent.mkdir(parents=True, exist_ok=True)
    create_output_link(src, dst)
    wait_for_real_nifti(dst)


# Clone dir: Colab=/content; local=WORKING_DIR/brainana_lite_env
ENV_DIR = Path("/content") if IS_COLAB else (WORKING_DIR / "brainana_lite_env")
BRAINANA_DIR = ENV_DIR / "brainana"
FIREANTS_DIR = ENV_DIR / "fireants"

def _installed(mod: str) -> bool:
    try:
        return importlib.util.find_spec(mod) is not None
    except Exception:
        return False

def _uv_cmd() -> list:
    """Return uv (bootstrap if missing). brainana needs uv for pinned deps + overrides."""
    uv = shutil.which("uv")
    if uv:
        return [uv]
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "uv"],
                   check=True, timeout=INSTALL_TIMEOUT)
    uv = shutil.which("uv")
    return [uv] if uv else [sys.executable, "-m", "uv"]


LITE_TEMPLATE_RES = "res-05"  # matches config template.output_space

# What a lite run actually needs out of the clone. Everything else is full-pipeline
# material: the Nextflow modules, the Docker build, the docs, the test suite and ~157 MB
# of example dataset. The filter used to exclude only template_zoo/ (122 MB) and then
# check out the remaining ~354 MB -- an allowlist brings that to ~80 MB.
_LITE_CODE_DIRS = ("src/",)
# README.md is not optional: pyproject.toml declares readme = {file = "README.md"}, so the
# build backend reads it during the editable install.
_LITE_CODE_FILES = ("pyproject.toml", "setup.cfg", "README.md", "LICENSE")
# ~85 MB of weights for skullstripping method "nhp_skullstrip_nn". Lite runs fastSurferCNN,
# whose own weights (src/fastsurfer_nn/pretrained_model) ARE fetched. These are declared in
# package-data as a glob, which simply matches nothing in an editable install.
_LITE_CODE_EXCLUDE = ("src/nhp_skullstrip_nn/pretrained_model/",)


def _lite_code_paths(listing: list[str]) -> list[str]:
    """Repo files outside template_zoo/ that a lite run needs (see the constants above)."""
    demo = f"examples/{DEMO_NIFTI}"
    return [
        p
        for p in listing
        if p
        and not p.startswith(_LITE_CODE_EXCLUDE)
        and (p.startswith(_LITE_CODE_DIRS) or p in _LITE_CODE_FILES or p == demo)
    ]

# fix_V1_WM templates are always under NMT2Sym (not TEMPLATE)
_NMT2SYM_FIX_V1_DIR = "template_zoo/template/NMT2Sym/"
_FASTSURFER_FIX_V1_TPL_SUFFIXES = (
    "tpl-NMT2Sym_res-05_T1w_brain.nii.gz",
    "tpl-NMT2Sym_res-05_T1w_WM_V1.nii.gz",
)
_FASTSURFER_FIX_V1_ATLAS_SUFFIX = "atlas-ARM2_space-NMT2Sym_res-05.nii.gz"


def _paths_with_suffix(listing: list[str], suffix: str, *, prefix: str = "") -> list[str]:
    return [p for p in listing if p.startswith(prefix) and p.endswith(suffix)]


def _lite_atlas_sidecar_paths(listing: list[str], atlas_paths: list[str]) -> list[str]:
    """Sidecars for the selected atlases: atlas-{name}.* (LUT/docs) + family references.bib.

    Mirrors copy_atlas_sidecars (steps/anatomical.py). These files carry no _space tag and
    are not .nii.gz, so the image-only atlas filter skips them; fetch them explicitly so
    copy_atlas_sidecars finds them on disk at runtime (the full zoo is never downloaded).
    """
    families: set[str] = set()
    prefixes: set[str] = set()
    for p in atlas_paths:
        family, _, base = p.rpartition("/")
        name = base.split("_space-")[0]  # e.g. atlas-ARM2
        families.add(family)
        prefixes.add(f"{family}/{name}.")
    out: list[str] = []
    for p in listing:
        family, _, base = p.rpartition("/")
        if base == "references.bib" and family in families:
            out.append(p)
        elif any(p.startswith(pref) for pref in prefixes) and not p.endswith((".nii.gz", ".json")):
            out.append(p)
    return out


def _lite_res_value(res_str: str | None) -> float:
    """Numeric value of a ``res-*`` token, mirroring utils.templates._parse_resolution.

    Deliberately reproduces that helper's arithmetic -- float("05") is 5.0, not 0.5 -- rather
    than the true millimetre value. The goal is to fetch exactly the file that
    TemplateManager.discover_atlases will later select at runtime; "correcting" the scale
    here would reintroduce the very mismatch this replaces.
    """
    if not res_str:
        return float("inf")
    res_str = str(res_str).strip().lower()
    if res_str.startswith("res-"):
        try:
            return float(res_str[4:])
        except ValueError:
            pass
    return float("inf")


def _lite_atlas_res(path: str) -> str | None:
    """The ``res-*`` token of an atlas path (utils.templates._extract_res_from_atlas_path)."""
    name = path.rsplit("/", 1)[-1]
    if name.endswith(".nii.gz"):
        name = name[: -len(".nii.gz")]
    for part in name.split("_"):
        if part.startswith("res-"):
            return part
    return None


def _lite_atlas_paths(listing: list[str], template: str, res: str) -> list[str]:
    """One atlas image per atlas in `template` space, chosen as the runtime chooses.

    Mirrors TemplateManager.discover_atlases: match by SPACE ONLY, then take the resolution
    closest to `res`. The old filter required an exact ``_space-{template}_{res}`` match, so
    every atlas stored at another resolution was silently never downloaded and quietly
    dropped from backprojection -- for TEMPLATE="D99" that discarded atlas-D99 itself.
    """
    space_tag = f"_space-{template}_"
    by_name: dict[str, list[str]] = {}
    for p in listing:
        base = p.rsplit("/", 1)[-1]
        if not p.startswith("template_zoo/atlas/") or not base.startswith("atlas-"):
            continue
        if space_tag not in base or not p.endswith(".nii.gz"):
            continue
        by_name.setdefault(base.split("_")[0], []).append(p)
    target = _lite_res_value(res)
    return [
        min(paths, key=lambda q: abs(_lite_res_value(_lite_atlas_res(q)) - target))
        for _, paths in sorted(by_name.items())
    ]


def _lite_template_zoo_paths(repo: Path, template: str, res: str) -> list[str]:
    """Paths under template_zoo/ to fetch for lite: one brain template + atlases in space."""
    listing = subprocess.run(
        ["git", "-C", str(repo), "ls-tree", "-r", "--name-only", "HEAD"],
        capture_output=True,
        text=True,
        check=True,
    ).stdout.splitlines()
    tpl_dir = f"template_zoo/template/{template}/"
    tpl_paths = [
        p
        for p in listing
        if p.startswith(tpl_dir) and res in p and p.endswith("_brain.nii.gz")
    ]
    atlas_paths = _lite_atlas_paths(listing, template, res)
    fix_v1_tpl_paths: list[str] = []
    for suffix in _FASTSURFER_FIX_V1_TPL_SUFFIXES:
        matches = _paths_with_suffix(listing, suffix, prefix=_NMT2SYM_FIX_V1_DIR)
        if not matches:
            matches = [p for p in listing if p.endswith(suffix)]
        fix_v1_tpl_paths.extend(matches)
    fix_v1_atlas = _paths_with_suffix(listing, _FASTSURFER_FIX_V1_ATLAS_SUFFIX, prefix="template_zoo/atlas/")
    if not fix_v1_atlas:
        fix_v1_atlas = [p for p in listing if p.endswith(_FASTSURFER_FIX_V1_ATLAS_SUFFIX)]
    if not tpl_paths:
        raise RuntimeError(
            f"No template brain volume found for {template}:{res} "
            f"(expected tpl-{template}_{res}_*_brain.nii.gz under {tpl_dir})."
        )
    all_tpl = tpl_paths + fix_v1_tpl_paths
    missing_fix_v1 = [
        s for s in _FASTSURFER_FIX_V1_TPL_SUFFIXES if not any(p.endswith(s) for p in all_tpl)
    ]
    if missing_fix_v1:
        raise RuntimeError(
            "fastsurfer_nn fix_V1_WM requires NMT2Sym templates missing from repo: "
            + ", ".join(missing_fix_v1)
            + f". Check BRAINANA_REF (current tree at {repo})."
        )
    atlas_sidecar_paths = _lite_atlas_sidecar_paths(listing, atlas_paths + fix_v1_atlas)
    return list(
        dict.fromkeys(
            tpl_paths + atlas_paths + atlas_sidecar_paths + fix_v1_tpl_paths + fix_v1_atlas
        )
    )


def _sparse_checkout_brainana_lite(repo: Path, ref: str, template: str, res: str) -> None:
    """Clone metadata only, then checkout code + lite template_zoo (not the full zoo)."""
    print(f"Cloning brainana ({ref}) — a few hundred MB, typically 1-3 min...", flush=True)
    subprocess.run(
        ["git", "clone", "--filter=blob:none", "--sparse", "--no-checkout", "--quiet",
         BRAINANA_URL, str(repo)],
        check=True,
        timeout=CLONE_TIMEOUT,
        # GIT_TERMINAL_PROMPT=0: fail fast on a bad URL / private repo instead of
        # blocking forever on a credential prompt no one can answer.
        env={**os.environ, "GIT_TERMINAL_PROMPT": "0"},
    )
    checkout = subprocess.run(
        ["git", "-C", str(repo), "checkout", "--quiet", ref],
        capture_output=True,
        text=True,
    )
    if checkout.returncode != 0:
        # check=True here raises a bare CalledProcessError naming only the git command,
        # which reads as a notebook bug rather than "that ref does not exist".
        raise RuntimeError(
            f"BRAINANA_REF={ref!r} was not found in {BRAINANA_URL}\n"
            f"  git: {(checkout.stderr or checkout.stdout).strip()}\n"
            "  Set BRAINANA_REF to a branch, tag or commit that exists in that repository.\n"
            "  A branch that only exists on someone's local machine is not visible here --\n"
            "  it has to be pushed first."
        )
    code_paths = _lite_code_paths(
        subprocess.run(
            ["git", "-C", str(repo), "ls-tree", "-r", "--name-only", "HEAD"],
            capture_output=True,
            text=True,
            check=True,
        ).stdout.splitlines()
    )
    zoo_paths = _lite_template_zoo_paths(repo, template, res)
    subprocess.run(
        ["git", "-C", str(repo), "sparse-checkout", "init", "--no-cone"],
        check=True,
    )
    subprocess.run(
        ["git", "-C", str(repo), "sparse-checkout", "set", "--stdin"],
        input="\n".join(code_paths + zoo_paths),
        text=True,
        check=True,
    )
    n_tpl = sum(1 for p in zoo_paths if p.startswith("template_zoo/template/"))
    n_atlas = sum(
        1 for p in zoo_paths
        if p.startswith("template_zoo/atlas/") and p.endswith(".nii.gz")
    )
    n_sidecar = len(zoo_paths) - n_tpl - n_atlas
    print(
        f"sparse checkout: {len(code_paths)} code file(s) + "
        f"{len(zoo_paths)} template_zoo file(s)",
        flush=True,
    )
    print(
        f"template_zoo (lite): {n_tpl} template volume(s) "
        f"({template}:{res} plus fastsurfer fix_V1), {n_atlas} atlas volume(s) "
        f"in {template} space, {n_sidecar} sidecar file(s)"
    )


# Idempotent: skip re-clone when ref + deps OK. FORCE_REINSTALL=True to refresh.
FORCE_REINSTALL = False


def _clone_is_current(repo: Path, ref: str) -> bool:
    """True if `repo` is a brainana clone whose HEAD is the commit `ref` names.

    Resolves `ref` to a commit instead of comparing strings. _sparse_checkout_brainana_lite
    checks out a tag, which leaves a DETACHED HEAD, so `rev-parse --abbrev-ref HEAD` returns
    the literal "HEAD" and can never equal a tag name -- the old string compare was always
    False for the pinned default, so every Run All re-cloned and reinstalled from scratch.
    Comparing commits works for tags, branches and SHAs alike.

    For a branch ref this compares against the *local* branch, so a clone is reused even if
    the remote has moved on; set FORCE_REINSTALL=True to refresh.
    """
    if not (repo / "src").exists():
        return False
    try:
        head = subprocess.run(
            ["git", "-C", str(repo), "rev-parse", "HEAD"],
            capture_output=True, text=True, check=True,
        ).stdout.strip()
        resolved = subprocess.run(
            ["git", "-C", str(repo), "rev-parse", "--verify", "--quiet", f"{ref}^{{commit}}"],
            capture_output=True, text=True, check=True,
        ).stdout.strip()
    except Exception:
        return False  # not a git repo / git missing / ref unknown -> treat as not current
    return bool(head) and head == resolved


step_log("Environment setup")
print(f"{'Colab' if IS_COLAB else 'Local'} run. WORKING_DIR: {WORKING_DIR}")
if ON_DRIVE_FUSE:
    print("  Google Drive detected — T1w inputs were sync-checked in the cell above.")
ENV_DIR.mkdir(parents=True, exist_ok=True)

# Env ready: matching clone ref + importable ants + torch
_env_ready = (
    not FORCE_REINSTALL
    and _clone_is_current(BRAINANA_DIR, BRAINANA_REF)
    and _installed("ants")
    and _installed("torch")
)

step_log("Clone and install brainana (uv)")
if _env_ready:
    step_log(
        "Clone and install brainana (uv)", done=True,
        detail=f"reused existing env (ref={BRAINANA_REF}); set FORCE_REINSTALL=True to refresh",
    )
else:
    if IS_COLAB:
        print(
            "First-time Colab setup (~few min). "
            "Runtime may restart once when install finishes — then Run all again."
        )
    if BRAINANA_DIR.exists():
        shutil.rmtree(BRAINANA_DIR)  # drop stale clone
    # Sparse checkout: code + template_zoo for TEMPLATE
    _sparse_checkout_brainana_lite(BRAINANA_DIR, BRAINANA_REF, TEMPLATE, LITE_TEMPLATE_RES)
    # uv install brainana[lite]; UV_TORCH_BACKEND picks torch wheel
    _torch_backend = "cpu" if sys.platform == "darwin" else "auto"
    _uv_env = {**os.environ, "UV_TORCH_BACKEND": _torch_backend}
    uv_extra = [] if (sys.prefix != sys.base_prefix) else ["--system"]
    # Resolve uv BEFORE the try: bootstrapping it shells out to pip, and a pip failure
    # raised inside the try would be misreported as "uv failed to install brainana[lite]".
    _uv = _uv_cmd()
    print(
        "Installing brainana[lite] with uv. On a fresh Colab runtime this downloads torch,\n"
        "antspyx and fireants (~2 GB) and takes several minutes. uv's output follows.",
        flush=True,
    )
    try:
        # On uv failure, read output above the traceback
        # cwd is load-bearing: it is how uv finds [tool.uv] override-dependencies in the
        # clone's pyproject.toml (fireants 1.5.0 hard-pins simpleitk==2.2.1). Dropping it
        # breaks resolution with an opaque "No solution found".
        # Streamed line by line rather than via run(): a child process inherits the kernel's
        # REAL stdout, which under Jupyter/Colab is not the cell output (IPython redirects
        # sys.stdout at the Python level only). Left to run(), uv prints nothing here and a
        # perfectly healthy multi-minute install is indistinguishable from a hang.
        _install = subprocess.Popen(
            [*_uv, "pip", "install", *uv_extra,
             "--python", sys.executable, "-e", f"{BRAINANA_DIR}[lite]"],
            cwd=str(BRAINANA_DIR), env=_uv_env,
            stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1,
        )
        # uv renders download progress as a \r-updated bar and suppresses it entirely when
        # stdout is a pipe, so between "Resolved N packages" and the final summary it emits
        # no newline-terminated output at all -- minutes of silence on a ~2 GB download.
        # A reader thread forwards whatever uv does print; the main loop adds a heartbeat so
        # the wait stays legible, and enforces INSTALL_TIMEOUT (Popen takes no timeout=).
        def _pump_uv_output() -> None:
            for _line in _install.stdout:
                print("    " + _line.rstrip(), flush=True)

        _pump = threading.Thread(target=_pump_uv_output, daemon=True)
        _pump.start()

        # uv stages every download in its cache, so the cache size is a progress proxy for
        # the phase uv itself reports nothing during. Growing => downloading; flat => building
        # a source distribution (fireants ships no wheels) or genuinely stuck. Without this the
        # heartbeat proves only that Python is alive, which is not the question being asked.
        _uv_cache = Path(os.environ.get("UV_CACHE_DIR") or Path.home() / ".cache" / "uv")

        def _cache_mb():
            """Size of uv's cache in MB, or None if it cannot be measured."""
            try:
                _du = subprocess.run(["du", "-sb", str(_uv_cache)],
                                     capture_output=True, text=True, timeout=15)
                return int(_du.stdout.split()[0]) / 1048576 if _du.returncode == 0 else None
            except Exception:
                return None

        _prev_mb = _cache_mb()
        # The flag keeps a timeout kill distinguishable from a genuine install failure --
        # otherwise it surfaces as the baffling "exit -9".
        _timed_out = []
        _started = time.monotonic()
        while True:
            try:
                _install.wait(timeout=INSTALL_HEARTBEAT)
                break
            except subprocess.TimeoutExpired:
                _elapsed = time.monotonic() - _started
                if _elapsed > INSTALL_TIMEOUT:
                    _timed_out.append(True)
                    _install.kill()
                    _install.wait()
                    break
                _now_mb = _cache_mb()
                _note = f"    ... still working ({_elapsed / 60:.1f} min elapsed"
                if _now_mb is not None:
                    _note += f", uv cache {_now_mb:,.0f} MB"
                    if _prev_mb is not None:
                        _delta = _now_mb - _prev_mb
                        _note += (f", +{_delta:,.0f} MB downloading" if _delta > 1
                                  else ", not growing -- building or stalled")
                    _prev_mb = _now_mb
                print(_note + ")", flush=True)
        _pump.join(timeout=5)
        if _timed_out:
            raise RuntimeError(
                f"uv exceeded INSTALL_TIMEOUT ({INSTALL_TIMEOUT}s) and was stopped. Re-run the\n"
                "cell -- uv caches what it already downloaded, so a retry resumes rather than\n"
                "starting over. Raise INSTALL_TIMEOUT if the connection is simply slow."
            )
        if _install.returncode != 0:
            raise subprocess.CalledProcessError(_install.returncode, _install.args)
    except subprocess.CalledProcessError as exc:
        raise RuntimeError(
            f"uv failed to install brainana[lite] (exit {exc.returncode}). The real cause is in\n"
            "uv's output printed just ABOVE this message - read it first. Common causes:\n"
            "  - A dependency has no wheel for this OS/CPU (look for 'No solution found' or\n"
            "    'no matching distribution'; e.g. torch has no Intel-macOS build - unsupported).\n"
            "  - A source build failed for missing system headers/compilers.\n"
            "  - Network/proxy trouble reaching PyPI."
        ) from exc
    step_log("Clone and install brainana (uv)", done=True, detail=f"ref={BRAINANA_REF}, dir={BRAINANA_DIR}")

# Ensure clone src/ is on sys.path (also after kernel restart)
_src_dir = str(BRAINANA_DIR / "src")
if _src_dir not in sys.path:
    sys.path.insert(0, _src_dir)
importlib.invalidate_caches()

# Restart kernel if uv upgraded binary deps (numpy/scipy/torch) already loaded in memory
import importlib.metadata as _ilm

_BINARY_DEPS = ("numpy", "scipy", "torch", "pandas", "matplotlib")
_stale_deps = []
for _dep in _BINARY_DEPS:
    _loaded_mod = sys.modules.get(_dep)
    if _loaded_mod is None:
        continue  # not yet imported -> will load the fresh on-disk version, no conflict
    _loaded_ver = getattr(_loaded_mod, "__version__", None)
    try:
        _disk_ver = _ilm.version(_dep)
    except _ilm.PackageNotFoundError:
        continue
    if _loaded_ver and _disk_ver and _loaded_ver != _disk_ver:
        _stale_deps.append(f"{_dep} {_loaded_ver}->{_disk_ver}")

if _stale_deps:
    _summary = ", ".join(_stale_deps)
    print("\n" + "=" * 72)
    print("Install complete. One restart so numpy/scipy/torch load cleanly.")
    print(f"  Updated: {_summary}")
    if IS_COLAB:
        print("  Restarting runtime… When Colab is back: Runtime → Run all.")
        print("  (Install is skipped on the second pass.)")
        print("=" * 72, flush=True)
        try:
            get_ipython().kernel.do_shutdown(restart=True)  # clean Colab runtime restart
        except Exception:
            os._exit(0)  # fallback: force the kernel to restart
        raise SystemExit("OK — restart session, then Run all again.")
    print("  Restart kernel (Kernel → Restart), then Run All.")
    print("=" * 72, flush=True)
    raise RuntimeError(
        "Install complete; restart the kernel, then Run All again (" + _summary + ")."
    )

# Optional FireANTs fused_ops (BUILD_FUSED_OPS=False; set True + nvcc for CUDA build)
BUILD_FUSED_OPS = False

step_log("FireANTs fused_ops (optional CUDA build)")
try:
    import torch
    if not BUILD_FUSED_OPS:
        step_log("FireANTs fused_ops (optional CUDA build)", done=True,
                 detail="skipped (BUILD_FUSED_OPS=False; non-fused GPU path used)")
    elif torch.cuda.is_available() and _installed("fireants"):
        if not FIREANTS_DIR.exists():
            subprocess.run(["git", "clone", "--quiet", "--branch", FIREANTS_REF, "--depth", "1",
                            FIREANTS_URL, str(FIREANTS_DIR)], check=True,
                           timeout=CLONE_TIMEOUT,
                           env={**os.environ, "GIT_TERMINAL_PROMPT": "0"})
        fused = FIREANTS_DIR / "fused_ops"
        if fused.exists():
            # check=False is deliberate (the run continues on the non-fused path), but the
            # outcome must be reported -- this used to log a green check mark on failure.
            _fused = subprocess.run(
                [sys.executable, str(fused / "setup.py"), "build_ext", "install"],
                cwd=str(fused), check=False, timeout=INSTALL_TIMEOUT)
            step_log("FireANTs fused_ops (optional CUDA build)", done=True,
                     detail="" if _fused.returncode == 0 else
                     f"BUILD FAILED (exit {_fused.returncode}); non-fused GPU path used")
        else:
            step_log("FireANTs fused_ops (optional CUDA build)", done=True, detail="fused_ops not found; skipped")
    else:
        step_log("FireANTs fused_ops (optional CUDA build)", done=True, detail="no CUDA or fireants; skipped")
except Exception as exc:
    step_log("FireANTs fused_ops (optional CUDA build)", done=True, detail=f"skipped ({exc})")

# Lite: antspyx only (no ANTs CLI)
if not _installed("ants"):
    raise RuntimeError(
        "Lite requires the antspyx (`ants`) package, but it is not importable.\n"
        "Ensure the brainana[lite] extra pulled in antspyx, then restart the kernel and Run All."
    )
print("Using antspyx backend for all ANTs ops (RIGID_METHOD=sitk).")


def _demo_raw_url() -> str:
    repo_slug = BRAINANA_URL.rstrip("/").removesuffix(".git").split("github.com/", 1)[-1]
    return f"https://raw.githubusercontent.com/{repo_slug}/{BRAINANA_REF}/examples/{DEMO_NIFTI}"


def _fetch_demo_t1w() -> Path:
    dst = WORKING_DIR / DEMO_NIFTI
    if dst.is_file():
        wait_for_real_nifti(dst)
        print(f"RUN_DEMO: using existing example at {dst}")
        return dst
    src = BRAINANA_DIR / "examples" / DEMO_NIFTI
    if src.is_file():
        shutil.copy2(src, dst)
        print(f"RUN_DEMO: copied example T1w from {src} -> {dst}")
    else:
        demo_url = _demo_raw_url()
        print(f"RUN_DEMO: downloading example T1w from {demo_url}")
        try:
            with urllib.request.urlopen(demo_url, timeout=DOWNLOAD_TIMEOUT) as resp, \
                    open(dst, "wb") as out:
                shutil.copyfileobj(resp, out)
        except Exception as exc:
            # Never leave a partial file: the Check T1w cell would read it as user input.
            dst.unlink(missing_ok=True)
            raise RuntimeError(
                f"RUN_DEMO failed to download {DEMO_NIFTI} from {demo_url}: {exc}"
            ) from exc
        print(f"RUN_DEMO: saved -> {dst}")
    wait_for_real_nifti(dst)
    print(f"  ✓ {dst.name}")
    return dst


if RUN_DEMO:
    step_log("RUN_DEMO — fetch example T1w")
    _fetch_demo_t1w()
    step_log("RUN_DEMO — fetch example T1w", done=True)

step_log("Environment setup", done=True)
```

```python
# Config + paths for pipeline steps
import yaml

config = load_config()

# load_config() deliberately does not validate, so a mistyped key used to create a brand-new
# entry that nothing reads -- the run then silently used the default the user believed they
# had overridden. Going through set_config() makes that a hard error instead.
_UNDECLARED_OK = {
    # Read via a plain-dict fallback in steps/anatomical.py (anat_conform), whose comment
    # names this notebook. Absent from defaults.yaml because the full pipeline uses flirt.
    "anat.conform.rigid_method",
}


def set_config(path: str, value) -> None:
    """Set a dotted config key, refusing to invent one that nothing reads."""
    *parents, leaf = path.split(".")
    node = config
    for key in parents:
        if not isinstance(node, dict) or key not in node:
            raise KeyError(f"config path {path!r}: {key!r} is not in defaults.yaml")
        node = node[key]
    if leaf not in node and path not in _UNDECLARED_OK:
        raise KeyError(f"config path {path!r}: {leaf!r} is not in defaults.yaml")
    node[leaf] = value


set_config("anat.surface_reconstruction.enabled", False)
set_config("anat.conform.rigid_method", "sitk")  # SimpleITK rigid conform: no FSL on Colab
set_config("template.output_space", f"{TEMPLATE}:{LITE_TEMPLATE_RES}")
set_config("general.gpu_device", "auto" if USE_GPU else -1)
set_config("general.verbose", 1 if VERBOSE else 0)  # align pipeline config with VERBOSE
# Dropped as no-ops: registration.fireants_allow_cpu (already defaults True) and
# general.anat_only (read only by BIDS discovery and main.nf, neither of which runs here).

effective_output_space = config["template"]["output_space"]
template_file = Path(resolve_template(effective_output_space))

modality = "T1w"
bids_stem = get_filename_stem(bids_name)
bids_prefix_wo_modality = bids_stem.replace(f"_{modality}", "")


def publish_copy(src: Path, filename: str) -> Path:
    dst = ANAT_DIR / filename
    dst.parent.mkdir(parents=True, exist_ok=True)
    create_output_link(src, dst)
    return dst


def _xfm_ext(path: Path) -> str:
    """Transform file extension (.h5, .nii.gz, etc.)."""
    resolved = Path(path).resolve()
    return "".join(resolved.suffixes) if resolved.suffixes else resolved.suffix


def show_qc(png_path: Path, title: str = "") -> None:
    if title:
        print(title)
    if png_path.exists():
        display(Image(filename=str(png_path)))
    else:
        print(f"QC image not found: {png_path}")


def qc_fig_path(desc: str) -> Path:
    """QC figure path, named as modules/qc.nf names it.

    Derived from the BIDS stem rather than a hand-assembled sub-/ses- prefix, so run- and
    acq- entities survive when the input is used as-is (no synthesis). Reads the current
    bids_name, which the synthesis step rebinds to the downstream naming template.
    """
    return FIG_DIR / create_bids_output_filename(
        bids_name, suffix=desc, modality=modality
    ).replace(".nii.gz", ".png")


def run_qc(qc_fn, save_f: Path, title: str = "", **kwargs) -> None:
    """Render one QC figure, reporting and continuing if it fails.

    The Docker pipeline runs every QC process under errorStrategy 'ignore', and the
    steps/qc.py wrappers catch exceptions, so a bad figure never costs a run. Calling the
    snapshot functions directly, as this notebook does, a single plotting error would abort
    the whole notebook *after* all the expensive compute had already succeeded.
    """
    try:
        qc_fn(save_f=str(save_f), **kwargs)
        show_qc(save_f, title)
    except Exception as exc:  # noqa: BLE001 - QC is diagnostic; never fail the run for it
        print(f"WARNING: QC figure skipped ({save_f.name}): {type(exc).__name__}: {exc}")

print(f"Template ({TEMPLATE}): {template_file}")

# BIDS derivatives root marker. Without it this output tree is not a valid derivatives
# dataset: no GeneratedBy, no brainana version, no template provenance at dataset level.
_dd_path = write_dataset_description(
    OUTPUT_DIR / "preproced",
    output_space=effective_output_space,
    resolved_template_path=str(template_file),
)
print(f"Dataset description: {_dd_path}")

# Run provenance. The Docker pipeline records the effective merged config under
# nextflow_reports/; without an equivalent, the configuration a Lite run actually used is
# recorded nowhere on disk. Best-effort: never fail a run over its own bookkeeping.
_report_dir = OUTPUT_DIR / "lite_reports"
try:
    _report_dir.mkdir(parents=True, exist_ok=True)
    (_report_dir / "config.yaml").write_text(
        yaml.safe_dump(
            {
                "brainana_lite": {
                    "brainana_ref": BRAINANA_REF,
                    "template": TEMPLATE,
                    "template_res": LITE_TEMPLATE_RES,
                    "resolved_template": str(template_file),
                    "use_gpu": USE_GPU,
                    "subject_id": SUBJECT_ID,
                    "session_id": SESSION_ID,
                },
                "config": config,
            },
            sort_keys=False,
            default_flow_style=False,
        )
    )
    print(f"Run provenance: {_report_dir / 'config.yaml'}")
except Exception as exc:  # noqa: BLE001 - bookkeeping must never abort a run
    print(f"WARNING: could not write run provenance: {exc}")
```

```python
# Step 1 — Synthesis (multi-run T1w within session)
step_log("Step 1 — Synthesis")
step_slow_hint()

synth_dir = WORK_ROOT / "synthesis"
synth_dir.mkdir(parents=True, exist_ok=True)

synth_out = anat_synthesis(anat_files=bids_files, working_dir=synth_dir, config=config)
synthesized = synth_out.metadata.get("synthesized", False)
is_subject_level = SESSION_ID is None

bids_output_filename, bids_name_downstream = create_synthesized_bids_filename(
    original_file=bids_name,
    modality=modality,
    is_subject_level=is_subject_level,
    synthesized=synthesized,
)

synth_file = synth_out.output_file
scanner_t1w_path = publish_copy(synth_file, bids_output_filename)

# Ingest normalization findings, derived exactly as modules/anatomical.nf derives them.
# anat_synthesis performs these repairs either way; recording them is the only trace that
# they happened. Every key is omitted when nothing fired, so a well-formed input still
# produces the sidecar it produced before.
_reports = synth_out.metadata.get("input_normalization") or []
_sidecar_extra = {"Synthesized": synthesized}
for _key, _field in (
    ("Input4DCollapsed", "dim"),
    ("OrientationRecovered", "orientation"),
    ("QformSformReconciled", "geometry"),
):
    _actions = [r[_field] for r in _reports if r.get(_field, "unchanged") != "unchanged"]
    if _actions:
        _sidecar_extra[_key] = _actions
_header_warnings = [w for r in _reports for w in r.get("warnings", [])]
if _header_warnings:
    _sidecar_extra["InputHeaderWarnings"] = _header_warnings

write_derivative_sidecar(
    scanner_t1w_path,
    skull_stripped=False,
    sources=[str(f) for f in bids_files],
    extra=_sidecar_extra,
)

if "OrientationRecovered" in _sidecar_extra:
    print(
        "WARNING: this input stored no orientation (qform_code == 0 and sform_code == 0), so\n"
        "         brainana assumed the nibabel/FSL convention. That recovers a CONVENTION,\n"
        "         not ground truth: if the acquisition ran the other way the result is a\n"
        "         left/right mirror that no registration can undo and no inspection reveals.\n"
        "         Confirm handedness against an external record before trusting any\n"
        "         hemisphere-wise result."
    )
if _header_warnings:
    print("WARNING: header defects with no safe automatic fix: " + "; ".join(_header_warnings))
step_log(
    "Step 1 — Synthesis",
    done=True,
    detail=f"synthesized={synthesized}, published={scanner_t1w_path.name}",
)
print(f"BIDS downstream template: {bids_name_downstream}")

# bids_name for steps 2+
bids_name = Path(bids_name_downstream)
bids_stem = get_filename_stem(bids_name)
bids_prefix_wo_modality = bids_stem.replace(f"_{modality}", "")
```

```python
# Step 2 — Conform to template
step_log("Step 2 — Conform to template")
step_slow_hint()
conform_dir = WORK_ROOT / "conform"
conform_dir.mkdir(parents=True, exist_ok=True)

conform_out = anat_conform(
    StepInput(
        input_file=synth_file,
        working_dir=conform_dir,
        config=config,
        output_name="anat_conformed.nii.gz",
        metadata=meta,
    ),
    template_file=template_file,
)

conformed_file = conform_out.output_file

# Full-FOV conform: the same conform on a grid enlarged so no scanner-space voxel is cropped
# (recording chamber, head-post, neck). A leaf output -- nothing downstream reads it -- but
# the only view of what the processing FOV cut away. anat_conform always attempts it; the
# trio is absent together when the helper falls back, so guard on all three.
full_fov_file = conform_out.additional_files.get("conformed_full_fov")
full_fov_template = conform_out.additional_files.get("template_resampled_full_fov")
fov_box_file = conform_out.additional_files.get("fov_box")
has_full_fov = bool(full_fov_file and full_fov_template and fov_box_file)

if full_fov_file:
    full_fov_pub = publish_copy(
        full_fov_file,
        create_bids_output_filename(
            bids_name, suffix=f"space-{modality}_desc-conformFullFOV", modality=modality
        ),
    )
    write_derivative_sidecar(
        full_fov_pub,
        skull_stripped=False,
        sources=[str(synth_file)],
        extra={"FullFOVPadding": conform_out.metadata.get("full_fov_pad")},
    )

template_resampled = conform_out.additional_files.get("template_resampled", template_file)
conform_qc_png = qc_fig_path("desc-conform")
_full_fov_qc_args = (
    dict(
        full_fov_file=str(full_fov_file),
        full_fov_template_file=str(full_fov_template),
        fov_box_file=str(fov_box_file),
        full_fov_save_f=str(qc_fig_path("desc-conformFullFOV")),
    )
    if has_full_fov
    else {}
)
run_qc(
    create_conform_qc,
    conform_qc_png,
    "Conform QC",
    conformed_file=str(conformed_file),
    template_file=str(template_resampled),
    modality="anat",
    logger=qc_logger,
    **_full_fov_qc_args,
)

# Publish conform affines
conform_xfm_published = []
_conform_fwd = conform_out.additional_files.get("forward_transform")
_conform_inv = conform_out.additional_files.get("inverse_transform")
if _conform_fwd:
    _p = publish_copy(
        _conform_fwd,
        f"{bids_prefix_wo_modality}_from-scanner_to-{modality}_mode-image_xfm.mat",
    )
    conform_xfm_published.append(_p.name)
    write_derivative_sidecar(_p, engine=conform_out.metadata.get("engine"), sources=[str(synth_file)])
if _conform_inv:
    _p = publish_copy(
        _conform_inv,
        f"{bids_prefix_wo_modality}_from-{modality}_to-scanner_mode-image_xfm.mat",
    )
    conform_xfm_published.append(_p.name)
    write_derivative_sidecar(_p, engine=conform_out.metadata.get("engine"), sources=[str(synth_file)])

_conform_detail = str(conformed_file)
if full_fov_file:
    _conform_detail += f", fullFOV={full_fov_pub.name}"
if conform_xfm_published:
    _conform_detail += f", xfm={', '.join(conform_xfm_published)}"
step_log("Step 2 — Conform to template", done=True, detail=_conform_detail)
```

```python
# Step 3 — Skull stripping + segmentation
step_log("Step 3 — Skull stripping + segmentation")
step_slow_hint()
skull_dir = WORK_ROOT / "skullstrip"
skull_dir.mkdir(parents=True, exist_ok=True)

skull_out = anat_skullstripping(
    StepInput(
        input_file=conform_out.output_file,
        working_dir=skull_dir,
        config=config,
        output_name="anat_brain.nii.gz",
        metadata=meta,
    )
)

brain_mask_src = skull_out.additional_files["brain_mask"]
seg_src = skull_out.additional_files.get("segmentation")
atlas_lut_src = skull_out.additional_files.get("atlas_lut")
hemimask_src = skull_out.additional_files.get("hemimask")
atlas_name = skull_out.metadata.get("atlas_name", "ARM2")

brain_stripped_file = skull_out.output_file
mask_path = publish_copy(brain_mask_src, f"{bids_prefix_wo_modality}_space-T1w_desc-brain_mask.nii.gz")
write_derivative_sidecar(mask_path, roi_type="Brain", sources=[str(conformed_file)])
if hemimask_src:
    # Published by the Docker pipeline whenever skullstripping returns it, which is always
    # for anat; roi_type="ROI" mirrors modules/anatomical.nf.
    hemimask_path = publish_copy(
        hemimask_src, f"{bids_prefix_wo_modality}_space-T1w_desc-brain_hemimask.nii.gz"
    )
    write_derivative_sidecar(hemimask_path, roi_type="ROI", sources=[str(conformed_file)])
if seg_src:
    seg_path = publish_copy(
        seg_src,
        f"{bids_prefix_wo_modality}_space-T1w_desc-brain_atlas{atlas_name}.nii.gz",
    )
    write_derivative_sidecar(seg_path, sources=[str(conformed_file)], extra={"Atlas": atlas_name})
if atlas_lut_src:
    lut_path = publish_copy(
        atlas_lut_src,
        f"{bids_prefix_wo_modality}_space-T1w_desc-brain_atlas{atlas_name}.tsv",
    )

skull_qc_png = qc_fig_path("desc-skullstrip")
run_qc(
    create_skullstripping_qc,
    skull_qc_png,
    "Skull stripping QC",
    underlay_file=str(conformed_file),
    mask_file=str(brain_mask_src),
    modality="anat",
    logger=qc_logger,
)
_step3_detail = f"mask={mask_path.name}"
if hemimask_src:
    _step3_detail += f", hemimask={hemimask_path.name}"
if seg_src:
    _step3_detail += f", seg={seg_path.name}"
if atlas_lut_src:
    _step3_detail += f", lut={lut_path.name}"
step_log("Step 3 — Skull stripping + segmentation", done=True, detail=_step3_detail)
```

```python
# Step 4 — Bias correction
step_log("Step 4 — Bias correction")
step_slow_hint()
bias_dir = WORK_ROOT / "bias"
bias_dir.mkdir(parents=True, exist_ok=True)

bias_out = anat_bias_correction(
    StepInput(
        input_file=conform_out.output_file,
        working_dir=bias_dir,
        config=config,
        output_name="anat_bias_corrected.nii.gz",
        metadata=meta,
    ),
    brain_mask=brain_mask_src,
)

bias_head_file = bias_out.output_file
if "brain" in bias_out.additional_files:
    bias_brain_file = bias_out.additional_files["brain"]
else:
    bias_brain_file = bias_out.output_file

bias_qc_png = qc_fig_path("desc-biascorrect")
run_qc(
    create_bias_correction_qc,
    bias_qc_png,
    "Bias correction QC",
    image_original=str(brain_stripped_file),
    image_corrected=str(bias_brain_file),
    modality="anat",
    logger=qc_logger,
)

if seg_src:
    seg_qc_png = qc_fig_path("desc-atlasSegmentation")
    run_qc(
        create_atlas_segmentation_qc,
        seg_qc_png,
        "Atlas segmentation QC",
        # Pre-bias brain, matching the Docker pipeline (anatomical_workflow.nf uses
        # anat_after_skull_brain here, not the bias-corrected image).
        underlay_file=str(brain_stripped_file),
        atlas_file=str(seg_path),
        modality="anat",
        logger=qc_logger,
    )

step_log(
    "Step 4 — Bias correction",
    done=True,
    detail=f"head={bias_head_file.name}, brain={bias_brain_file.name}",
)
```

```python
# Publish desc-preproc T1w outputs
step_log("Publish Phase 1 (T1w desc-preproc)")
preproc_head = publish_copy(
    bias_head_file,
    create_bids_output_filename(bids_name, suffix="space-T1w_desc-preproc", modality=modality),
)
preproc_brain = publish_copy(
    bias_brain_file,
    f"{bids_prefix_wo_modality}_space-T1w_desc-preproc_{modality}_brain.nii.gz",
)
write_derivative_sidecar(preproc_head, skull_stripped=False, sources=[str(bias_head_file)])
write_derivative_sidecar(preproc_brain, skull_stripped=True, sources=[str(bias_brain_file)])

step_log("Publish Phase 1 (T1w desc-preproc)", done=True, detail=f"{preproc_head.name}, {preproc_brain.name}")
```

```python
# Step 5 — Registration (brain -> template)
step_log("Step 5 — Registration (brain -> template)")
step_slow_hint()
reg_dir = WORK_ROOT / "registration"
reg_dir.mkdir(parents=True, exist_ok=True)

# input: bias-corrected brain
reg_brain_out = anat_registration(
    StepInput(
        input_file=bias_brain_file,
        working_dir=reg_dir,
        config=config,
        output_name="anat_registered_brain.nii.gz",
        metadata=meta,
    ),
    template_file=template_file,
    template_name=TEMPLATE,
)

forward_xfm = reg_brain_out.additional_files["forward_transform"]
inverse_xfm = reg_brain_out.additional_files["inverse_transform"]

# Publish registration transforms
reg_xfm_published = []
fwd_xfm_path = publish_copy(
    forward_xfm,
    f"{bids_prefix_wo_modality}_from-{modality}_to-{TEMPLATE}_mode-image_xfm{_xfm_ext(forward_xfm)}",
)
reg_xfm_published.append(fwd_xfm_path.name)
inv_xfm_path = publish_copy(
    inverse_xfm,
    f"{bids_prefix_wo_modality}_from-{TEMPLATE}_to-{modality}_mode-image_xfm{_xfm_ext(inverse_xfm)}",
)
reg_xfm_published.append(inv_xfm_path.name)
for _xfm_pub in (fwd_xfm_path, inv_xfm_path):
    write_derivative_sidecar(
        _xfm_pub,
        output_space=effective_output_space,
        resolved_template_path=str(template_file),
        engine=reg_brain_out.metadata.get("engine"),
        sources=[str(bias_brain_file), str(template_file)],
    )
step_log(
    "Step 5 — Registration (brain -> template)",
    done=True,
    detail=f"xfm={', '.join(reg_xfm_published)}",
)

step_log("Step 5 — Apply registration to full head")
interpolation = config.get("registration", {}).get("interpolation", "BSpline")
apply_result = ants_apply_transforms(
    movingf=str(bias_head_file),
    moving_type=0,
    interpolation=interpolation,
    outputf_name="anat_registered_head.nii.gz",
    fixedf=str(template_file),
    working_dir=str(reg_dir / "apply_head"),
    transformf=[str(forward_xfm)],
    reff=str(template_file),
    logger=qc_logger,
)

registered_head = Path(apply_result["imagef_registered"])
reg_head_path = publish_copy(
    registered_head,
    create_bids_output_filename(
        bids_name,
        suffix=f"space-{TEMPLATE}_desc-preproc",
        modality=modality,
    ),
)
write_derivative_sidecar(
    reg_head_path,
    output_space=effective_output_space,
    resolved_template_path=str(template_file),
    skull_stripped=False,
    sources=[str(bias_head_file)],
)

reg_qc_png = qc_fig_path("desc-anat2template")
run_qc(
    create_registration_qc,
    reg_qc_png,
    "Registration QC",
    image_file=str(reg_head_path),
    template_file=str(template_file),
    modality="anat2template",
    logger=qc_logger,
)
step_log("Step 5 — Apply registration to full head", done=True, detail=str(reg_head_path))
```

```python
# Apply brain mask to template space
step_log("Apply brain mask to template space")
mask_reg_dir = WORK_ROOT / "mask_to_template"
mask_reg_dir.mkdir(parents=True, exist_ok=True)

mask_apply = ants_apply_transforms(
    movingf=str(brain_mask_src),
    moving_type=0,
    interpolation="NearestNeighbor",
    outputf_name="mask_registered.nii.gz",
    fixedf=str(template_file),
    working_dir=str(mask_reg_dir),
    transformf=[str(forward_xfm)],
    reff=str(template_file),
    logger=qc_logger,
    generate_tmean=False,
)

mask_registered = Path(mask_apply["imagef_registered"])
mask_template_name = f"{replace_bids_space(bids_prefix_wo_modality, TEMPLATE)}_desc-brain_mask.nii.gz"
mask_template_path = publish_copy(mask_registered, mask_template_name)
write_derivative_sidecar(
    mask_template_path,
    output_space=effective_output_space,
    resolved_template_path=str(template_file),
    roi_type="Brain",
    # mask + transform + reference, matching modules/anatomical.nf
    sources=[str(brain_mask_src), str(forward_xfm), str(template_file)],
)
step_log("Apply brain mask to template space", done=True, detail=str(mask_template_path))
```

```python
# Step 6 — Atlas backproject to T1w space
step_log("Step 6 — Atlas backproject to T1w space")
atlas_t1w_dir = WORK_ROOT / "atlas_t1w"
atlas_t1w_dir.mkdir(parents=True, exist_ok=True)

bp_out = anat_backproject_atlases(
    inverse_xfm=inverse_xfm,
    t1w_reference=preproc_head,
    bids_name=bids_name,
    working_dir=atlas_t1w_dir,
    config=config
)

atlas_t1w_pub = ANAT_DIR / "atlas_space-T1w"
atlas_t1w_pub.mkdir(parents=True, exist_ok=True)
atlas_t1w_files = []
for atlas_name_key, atlas_path in bp_out.additional_files.items():
    dst = atlas_t1w_pub / atlas_path.name
    shutil.copy2(atlas_path, dst)
    atlas_t1w_files.append(dst)
    print(f"  T1w atlas: {dst.name}")

# Publish the atlas LUT/docs/refs the step copied into its work dir
# (atlas-{name}.tsv/.md/.bib), mirroring the Docker pipeline's atlas_space-T1w
# publish. bp_out.output_file is that work atlas/ dir; sidecars are space-independent.
for _sc in sorted(bp_out.output_file.glob("atlas-*.*")):
    if _sc.suffix in {".tsv", ".md", ".bib"}:
        shutil.copy2(_sc, atlas_t1w_pub / _sc.name)
        print(f"  T1w atlas sidecar: {_sc.name}")

# Write the per-atlas JSON provenance sidecar next to each published label volume,
# matching the Docker pipeline (which generates these via write_derivative_sidecar,
# not copy_atlas_sidecars). Same ROI type and source transforms as the .nf process.
for _atlas_pub in atlas_t1w_files:
    write_derivative_sidecar(
        _atlas_pub,
        roi_type="ROI",
        sources=[str(inverse_xfm), str(preproc_head)],
    )

step_log("Step 6 — Atlas backproject to T1w space", done=True, detail=f"{len(atlas_t1w_files)} atlas(es)")
```

```python
# Step 7 — Atlas to scanner space
step_log("Step 7 — Atlas reprojection to scanner space")
atlas_scanner_dir = WORK_ROOT / "atlas_scanner"
atlas_scanner_dir.mkdir(parents=True, exist_ok=True)

conform_inverse = conform_out.additional_files.get("inverse_transform")
if conform_inverse is None:
    raise FileNotFoundError("Conform inverse transform not found")

scanner_out = anat_reproject_atlases_to_scanner(
    atlas_files=atlas_t1w_files,
    conform_inverse_xfm=conform_inverse,
    scanner_reference=synth_file,
    working_dir=atlas_scanner_dir,
    rigid_method="sitk",
)

atlas_scanner_pub = ANAT_DIR / "atlas_space-scanner"
atlas_scanner_pub.mkdir(parents=True, exist_ok=True)
atlas_scanner_files = []
for _, atlas_path in scanner_out.additional_files.items():
    dst = atlas_scanner_pub / Path(atlas_path).name
    shutil.copy2(atlas_path, dst)
    atlas_scanner_files.append(dst)
    print(f"  Scanner atlas: {dst.name}")

# Publish atlas sidecars into scanner space too (same LUT/docs/refs; space-independent),
# mirroring the Docker pipeline's atlas_space-scanner publish.
for _sc in sorted(scanner_out.output_file.glob("atlas-*.*")):
    if _sc.suffix in {".tsv", ".md", ".bib"}:
        shutil.copy2(_sc, atlas_scanner_pub / _sc.name)
        print(f"  Scanner atlas sidecar: {_sc.name}")

# JSON provenance sidecar per scanner-space atlas (matches the .nf process:
# roi_type ROI, sourced from the conform inverse transform and scanner reference).
for _atlas_pub in atlas_scanner_files:
    write_derivative_sidecar(
        _atlas_pub,
        roi_type="ROI",
        sources=[str(conform_inverse), str(synth_file)],
    )

step_log("Step 7 — Atlas reprojection to scanner space", done=True)
```
